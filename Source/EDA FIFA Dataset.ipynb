```python
# Verificando si existen filas totalmente duplicadas
print(f"Filas duplicadas encontradas: {df.duplicated().sum()}")
```

```python
# 1. Crear Dimensión de Jugadores (1 fila por jugador)
cols_player = ['player_id', 'player_name', 'age', 'nationality', 'position', 
               'height_cm', 'weight_kg', 'preferred_foot', 'club_name', 'market_value_eur']
df_player = df[cols_player].drop_duplicates(subset=['player_id']).copy()

# 2. Crear Dimensión de Partidos (1 fila por partido)
cols_match = ['match_id', 'match_date', 'stadium', 'city', 'tournament_stage']
df_match = df[cols_match].drop_duplicates(subset=['match_id']).copy()

# 3. Crear Tabla de Hechos (Métricas y llaves foráneas por partido/jugador)
cols_fact = ['player_id', 'match_id', 'team', 'opponent_team', 'minutes_played', 
             'goals', 'assists', 'shots', 'expected_goals_xg', 'key_passes', 
             'successful_passes', 'total_passes', 'pass_accuracy', 'tackles', 
             'interceptions', 'yellow_cards', 'red_cards', 'player_rating']
df_fact = df[cols_fact].copy()
```

```python
# Guardar usando rutas absolutas en tu carpeta Data
df_player.to_csv(r"C:\Users\faust\Desktop\Personal\Cursos\Curso PBI\Proyectos\Proyecto 2 - FIFA 2026\Dim_Player.csv", index=False)
df_match.to_csv(r"C:\Users\faust\Desktop\Personal\Cursos\Curso PBI\Proyectos\Proyecto 2 - FIFA 2026\Dim_Match.csv", index=False)
df_fact.to_csv(r"C:\Users\faust\Desktop\Personal\Cursos\Curso PBI\Proyectos\Proyecto 2 - FIFA 2026\Fact_Performance.csv", index=False)

print("¡Tablas dimensionales y de hechos generadas y exportadas con éxito!")
```
